```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding="latin1")
```

```python
df.head()
```

```python
df.info()
```

```python
df = df.drop("Customer Email", axis=1)
```

```python
df = df.drop("Customer Password", axis=1)
```

```python
df.rename(columns={
    "order date (dateorders)": "order date"
}, inplace=True)
```

```python
df["order date"] = pd.to_datetime(df["order date"])
```

```python
df.rename(columns={
    "shipping date (dateorders)": "shipping date"
}, inplace=True)
```

```python
df["shipping date"] = pd.to_datetime(df["shipping date"])
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace("_", " ")
```

```python
df.info()
```

```python
df = df.drop("product description", axis=1)
```

```python
df = df.drop("product status", axis=1)
```

```python
df.to_csv("dataco.csv", index=False)
```
